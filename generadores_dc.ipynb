# **¿Qué es un motor DC y cómo funciona?**

**Un motor DC** (corriente continua) es un dispositivo que transforma la energía eléctrica en energía mecánica, produciendo un movimiento giratorio.

## **¿Cómo funciona?**
Cuando conectamos una pila al motor, la corriente eléctrica circula por sus bobinas y genera un campo magnético que interactúa con los imanes del motor, haciendo girar su eje.

## **¿Puede generar electricidad?**
¡Sí! Algunos motores DC también pueden funcionar como generadores eléctricos.

Cuando hacemos girar su eje mediante el viento, el agua o nuestra mano, el movimiento produce electricidad que podemos utilizar para encender un LED.

**En resumen:**
- Como motor: electricidad → movimiento.
- Como generador: movimiento → electricidad.
Este es el principio que utilizamos en nuestro proyecto «Mini Generadores Eléctricos», donde aprovechamos la fuerza del viento y del agua para producir electricidad.

**El tamaño del motor influye en la cantidad de electricidad que puede generar, pero un motor más grande no necesariamente genera más electricidad que uno pequeño.**

**¿Qué diferencias hay entre un motor grande y uno pequeño?**

|Característica	|Motor pequeño	|Motor grande|
|:-:|:-:|:-:|
|Peso	|Menor	|Mayor|
|Resistencia al giro	|Depende del diseño	|Depende del diseño|
|Voltaje generado	|Depende de las RPM y del motor	|Depende de las RPM y del motor|
|Corriente que puede entregar	|Generalmente menor	|Potencialmente mayor|
|Facilidad para moverlo con viento	|Puede ser más fácil	|Puede requerir más fuerza|

# **¿Qué determina cuánta electricidad genera?**

Hay tres factores fundamentales:

**Velocidad de giro (RPM).** Generalmente, cuanto más rápido gira un motor DC utilizado como generador, mayor voltaje produce.

**Características del motor.** Algunos motores producen más voltaje por cada revolución que otros.

**Fuerza disponible (torque).** Para generar corriente y alimentar un LED, el molino necesita suficiente fuerza para mantener girando el motor.

Por eso, un generador que funciona perfectamente cuando lo girás con la mano podría no encender el LED cuando lo mueve el viento.
